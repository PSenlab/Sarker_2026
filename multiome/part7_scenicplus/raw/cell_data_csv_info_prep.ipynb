```python
import scanpy as sc

# Assuming `adata` contains the clustering information
# Extract the relevant columns: 'sample', 'celltype', and the Leiden clustering results
cell_info = GEX_anndata.obs[['sample', 'sex', 'age', 'celltype', 'leiden_1', 'leiden_2', 'leiden_3', 'leiden_4', 'leiden_5', 'celltype2']].copy()

# Extract the barcode from the index
cell_info['barcode'] = cell_info.index

# Ensure that the barcode is correctly formatted (without the sample suffix but with the "-1" suffix)
cell_info['barcode'] = cell_info['barcode'].apply(lambda x: '-'.join(x.split('-')[:2]))

# Reorder columns to have barcode first
columns_order = ['barcode', 'sample', 'sex', 'age', 'celltype', 'leiden_1', 'leiden_2', 'leiden_3', 'leiden_4', 'leiden_5', 'celltype2']
cell_info = cell_info[columns_order]

# Display the first few rows of the DataFrame
print(cell_info.head())

# Save the DataFrame to a TSV file
cell_info.to_csv("cell_data.tsv", sep='\t', index=False)
```
